```python
import csv
from collections import defaultdict
from pathlib import Path
rows = list(csv.DictReader((Path.cwd() / "superstore-summary.csv").open()))
sales = defaultdict(float)
for row in rows:
    sales[row["Region"]] += float(row["Sales"])
for region, value in sorted(sales.items(), key=lambda x: -x[1]):
    print(f"{region}: ${value:,.2f}")
```
